```python
import pandas as pd
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["target"] = iris.target
df["target_name"] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

# Display first 10 rows
print("\nFirst 10 Rows:")
print(df.head(10))

# ============================================================
# 1. SELECT COLUMNS USING LOC
# ============================================================

print("\n--- Column Selection using loc ---")

loc_columns = df.loc[:, ["sepal length (cm)", "sepal width (cm)", "target_name"]]

print(loc_columns.head(10))

# ============================================================
# 2. SELECT COLUMNS USING ILOC
# ============================================================

print("\n--- Column Selection using iloc ---")

iloc_columns = df.iloc[:, [0, 1, 4, 5]]

print(iloc_columns.head(10))

# ============================================================
# 3. SELECT ROWS USING LOC
# ============================================================

print("\n--- Row Selection using loc ---")

rows_loc = df.loc[0:9]

print(rows_loc)

# ============================================================
# 4. SELECT ROWS USING ILOC
# ============================================================

print("\n--- Row Selection using iloc ---")

rows_iloc = df.iloc[0:10]

print(rows_iloc)

# ============================================================
# 5. SELECT SPECIFIC ROWS AND COLUMNS USING LOC
# ============================================================

print("\n--- Specific Rows and Columns using loc ---")

specific_loc = df.loc[
    0:9,
    ["sepal length (cm)", "petal length (cm)", "target_name"]
]

print(specific_loc)

# ============================================================
# 6. SELECT SPECIFIC ROWS AND COLUMNS USING ILOC
# ============================================================

print("\n--- Specific Rows and Columns using iloc ---")

specific_iloc = df.iloc[0:10, [0, 2, 4, 5]]

print(specific_iloc)

# ============================================================
# 7. CONDITIONAL SELECTION
# ============================================================

print("\n--- Conditional Selection ---")

condition = df[df["sepal length (cm)"] > 6.0]

print(condition.head(10))

# ============================================================
# 8. MULTIPLE CONDITIONS
# ============================================================

print("\n--- Multiple Conditions ---")

multiple_conditions = df[
    (df["sepal length (cm)"] > 6.0) &
    (df["petal length (cm)"] > 4.0)
]

print(multiple_conditions.head(10))

# ============================================================
# 9. SELECT ONLY REQUIRED COLUMNS AFTER CONDITION
# ============================================================

print("\n--- Conditional Selection with Required Columns ---")

result = df.loc[
    df["sepal length (cm)"] > 6.0,
    ["sepal length (cm)", "petal length (cm)", "target_name"]
]

print(result.head(10))

# ============================================================
# 10. FINAL DATASET INFORMATION
# ============================================================

print("\n--- Dataset Information ---")

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nDataset Summary:")
print(df.describe())

print("\nTarget Class Distribution:")
print(df["target_name"].value_counts())
```
